```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np
import os

print("=== INICIO DE GENERACIÓN DE GRÁFICO ULTRA-VISUAL ===")

try:
    # --- CONFIGURACIÓN DE ARCHIVOS ---
    # Usaremos el archivo v2 limpio para garantizar que la columna 'SIGNO_ZODIACAL' esté bien
    archivo_csv = 'v2_dataset_estadisticamente_limpio.csv'
    columna_signo = 'SIGNO_ZODIACAL' 
    nombre_archivo_salida = '../Results/v0_distribucion_signo_zodiacal_ULTRA.png'
    
    # Carga de datos
    df_final = pd.read_csv(archivo_csv)
    conteo_signos = df_final[columna_signo].value_counts().sort_index()
    
    # --- GRÁFICO DE BARRAS MEJORADO ---
    plt.style.use('default') # Resetear estilo para máxima claridad
    fig, ax = plt.subplots(figsize=(15, 8))
    
    barras = conteo_signos.plot(kind='bar', ax=ax, color='mediumpurple', edgecolor='black', alpha=0.9)
    
    # 1. TÍTULOS Y EJES
    ax.set_title('Distribución de Pacientes por Signo Zodiacal (N > 5.8 Millones)', fontsize=20, fontweight='bold', pad=20)
    ax.set_xlabel('Signo Zodiacal', fontsize=15, fontweight='bold', labelpad=15)
    ax.set_ylabel('Número de Pacientes', fontsize=15, fontweight='bold')
    
    # Rotar etiquetas X y ajustar fuente
    plt.xticks(rotation=45, ha='right', fontsize=11)
    
    # 2. ETIQUETAS DE VALOR (Numeros exactos sobre las barras)
    for bar in barras.patches:
        # Altura de la barra
        yval = bar.get_height()
        # Formatear el número con comas y sin decimales
        ax.text(bar.get_x() + bar.get_width()/2, yval + 10000, 
                f'{yval:,.0f}', ha='center', va='bottom', fontsize=10, fontweight='semibold')

    # 3. FORMATO DEL EJE Y (Mostrar en K o M para limpieza visual)
    def millions_formatter(x, pos):
        return f'{x*1e-6:.1f} M'

    # Se usa un formato más sencillo de números enteros para no perder precisión en el eje
    ax.get_yaxis().set_major_formatter(ticker.FuncFormatter(lambda x, pos: f'{x:,.0f}'))

    # 4. AÑADIR LÍNEA DE PROMEDIO (Para resaltar la uniformidad)
    promedio = conteo_signos.mean()
    ax.axhline(promedio, color='red', linestyle='--', linewidth=2, label=f'Promedio: {promedio:,.0f}')
    ax.text(0.98, promedio + 15000, f'Media', color='red', ha='right', va='center', fontsize=12, fontweight='bold')

    # Otros ajustes
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    ax.ticklabel_format(style='plain', axis='y') # Asegura que no use notación científica
    plt.legend(loc='lower left')
    plt.tight_layout()
    
    # Guardar el gráfico final
    os.makedirs(os.path.dirname(nombre_archivo_salida), exist_ok=True)
    plt.savefig(nombre_archivo_salida, dpi=300)
    
    print(f"\n✅ Gráfico de distribución ULTRA-VISUAL guardado como '{nombre_archivo_salida}'.")

except FileNotFoundError:
    print(f"❌ Error: El archivo '{archivo_csv}' no fue encontrado. Revisa la ruta.")
except KeyError:
    print(f"❌ Error: La columna '{columna_signo}' no fue encontrada. Revisa el nombre exacto.")
except Exception as e:
    print(f"❌ Ocurrió un error inesperado al generar el gráfico: {str(e)}")
```
